```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import pandas as pd

# 1. Separate features (X) from the target variable (y)
X = df.drop('income', axis=1)
y = df['income']

# 2. Standardize the full feature set
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Apply PCA (retaining 95% of the variance)
pca = PCA(n_components=0.95, random_state=42)
X_pca = pca.fit_transform(X_scaled)

print(f"Original number of features: {X.shape[1]}")
print(f"Reduced number of features (95% variance): {X_pca.shape[1]}")

# 4. Create a new DataFrame with the reduced principal components
pca_columns = [f'PC{i+1}' for i in range(X_pca.shape[1])]
df_pca = pd.DataFrame(X_pca, columns=pca_columns)

# Re-attach the target variable for future modeling
df_pca['income'] = y.values

df_pca.head()
```

```python
x = df_pca[['PC1', 'PC2']]
x.head()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

wcss = []

for i in range(2, 11):
    km_object = KMeans(
        n_clusters=i,
        init='k-means++',
        max_iter=300,
        n_init=10,
        random_state=0
    )
    km_object.fit(x)
    wcss.append(km_object.inertia_)
    score = silhouette_score(x, km_object.labels_)
    print(f'{i} clusters - Silhouette Score: {score:.4f}')
```

```python
# Plot Elbow curve
plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), wcss, c='red', marker='o')
plt.title('Elbow Method (PC1 vs PC2)')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.grid(True)
plt.show()
```

```python
wcss
```

```python
km_object = KMeans(
    n_clusters=4,
    init='k-means++',
    max_iter=300,
    n_init=10,
    random_state=0
)
km_object.fit(x)

kmeans_cluster_labels = km_object.labels_
kmeans_cluster_labels
```

```python
df['kmeans_cluster'] = kmeans_cluster_labels
df.head()
```

```python
df_pca['kmeans_cluster'] = kmeans_cluster_labels
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df_pca,
    x='PC1',
    y='PC2',
    hue='kmeans_cluster',
    palette='Set1',
    alpha=0.6
)
plt.title('K-Means Clusters on First Two Principal Components')
plt.xlabel('Principal Component 1 (PC1)')
plt.ylabel('Principal Component 2 (PC2)')
plt.show()
```

```python
# 1. Attach cluster labels to original and PCA data
df['kmeans_cluster'] = df_pca['kmeans_cluster'].values

# 2. Inspect cluster centroids in PCA space
print("--- PCA Space Centroids ---")
print(df_pca.groupby('kmeans_cluster')[['PC1', 'PC2']].mean())

# 3. Inspect demographic breakdown across clusters
print("\n--- Original Feature Means per Cluster ---")
print(df.groupby('kmeans_cluster')[['age', 'hours-per-week', 'education-num', 'income']].mean())
```

Hierarchial Clustering - Agglomerative Clusering

```python
from scipy.cluster.hierarchy import linkage, dendrogram
from sklearn.cluster import AgglomerativeClustering
```

```python
# Subset for hierarchical clustering
x_sample = x.sample(n=2000, random_state=42)

# Dendrogram
mergings = linkage(x_sample, method='ward')
plt.figure(figsize=(10, 6))
dendrogram(mergings)
plt.title('Dendrogram (PC1 vs PC2)')
plt.xlabel('Sample Index')
plt.ylabel('Euclidean Distance')
plt.show()
```

```python
k_optimal = 4 # Based on K-Means elbow method and silhouette scores

# Fit Agglomerative Clustering model
ac = AgglomerativeClustering(n_clusters=k_optimal, metric='euclidean', linkage='ward')
ac_labels = ac.fit_predict(x_sample)

# Compare Silhouette scores on the exact same sample
ac_score = silhouette_score(x_sample, ac_labels)
km_score = silhouette_score(x_sample, df_pca.loc[x_sample.index, 'kmeans_cluster'])

print(f"Agglomerative Silhouette Score: {ac_score:.4f}")
print(f"K-Means Silhouette Score:       {km_score:.4f}")
```

```python
df_sample = x_sample.copy()
df_sample['ac_clusters'] = ac_labels
df_sample['kmeans_cluster'] = df_pca.loc[x_sample.index, 'kmeans_cluster']

plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df_sample,
    x='PC1',
    y='PC2',
    hue='ac_clusters',
    palette='tab10',
    alpha=0.7
)
plt.title('Agglomerative Hierarchical Clustering (PC1 vs PC2)')
plt.xlabel('Principal Component 1 (PC1)')
plt.ylabel('Principal Component 2 (PC2)')
plt.show()
```

```python
# 3. Check cluster distributions
print("--- Agglomerative Cluster Counts (Sample of 2000) ---")
print(df_sample['ac_clusters'].value_counts())
```

```python
print("\n--- K-Means Cluster Counts (Full Dataset) ---")
print(df_pca['kmeans_cluster'].value_counts())
```

```python
# 4. Compare Silhouette Scores (calculated on the same sample for fair comparison)
score_ac = silhouette_score(x_sample, df_sample['ac_clusters'])
score_km = silhouette_score(x_sample, df_sample['kmeans_cluster'])

print(f"\nSilhouette Score (Agglomerative): {score_ac:.4f}")
print(f"Silhouette Score (K-Means):        {score_km:.4f}")
```
